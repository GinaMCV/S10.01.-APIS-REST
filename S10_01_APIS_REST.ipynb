# Nivel 1
Desde un Jupyter Notebook realizarás los siguientes ejercicios utilizando la librería requests de Python.
Exploración básica con una API de laboratorio.

Recursos:
- [{JSON} Placeholder](https://jsonplaceholder.typicode.com)


```python
BASE_URL = "https://jsonplaceholder.typicode.com"
```

## 1. Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:
- Lista de publicaciones (/posts)
- Lista de usuarios (/users)
- Lista de tareas (/todos)

## 2. Muestra por pantalla:

- La cantidad total de cada recurso.
- El código de estado de cada petición.

```python
datos = {}
for recurso in ["posts", "users", "todos"]:
    resp = requests.get(f"{BASE_URL}/{recurso}", timeout=10)
    datos[recurso] = resp.json()
    print(f"/{recurso:<6} -> código de estado: {resp.status_code} | total de registros: {len(datos[recurso])}")
```

## 3. Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

```python
resp = requests.get(f"{BASE_URL}/posts/101", timeout=10)
print("Código de estado:", resp.status_code)
```

## 4. Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.

- Muestra la respuesta JSON.
- Muestra el código de estado.

```python
nueva_publicacion = {
    "title": "Publicacion de prueba",
    "body": "Este es el contenido de una publicación creada con como prueba.",
    "userId": 1,
}

resp = requests.post(f"{BASE_URL}/posts", json=nueva_publicacion, timeout=10)
print("Código de estado:", resp.status_code)
print("Respuesta JSON:")
print(resp.json())
```

## 5. Realiza una petición PATCH para modificar parcialmente una publicación existente.

- Muestra la respuesta JSON.
- Muestra el código de estado.

```python
cambios = {"title": "Nuevo título modificado"}

resp = requests.patch(f"{BASE_URL}/posts/1", json=cambios, timeout=10)
print("Código de estado:", resp.status_code)
print("Respuesta JSON:")
print(resp.json())
```

## 6. Haz una petición DELETE sobre una publicación.

- Muestra la respuesta JSON.
- Muestra el código de estado.

```python
resp = requests.delete(f"{BASE_URL}/posts/1", timeout=10)
print("Código de estado:", resp.status_code)
print("Respuesta JSON:", resp.json())
```

## 3. Haz una petición GET sencilla:

- Muestra el código de estado de la respuesta.
- Imprime de forma clara algunos de los campos de la respuesta JSON.

```python
BASE = "https://ghibliapi.vercel.app"

resp = requests.get(f"{BASE}/films", timeout=10)
print("Código de estado:", resp.status_code)
print("Content-Type:", resp.headers.get("Content-Type"))

films = resp.json()
print("Número de películas:", len(films))
print("Campos disponibles:", list(films[0].keys()))

print("\nAlgunas películas:")
for f in films[:5]:
    print(f"- {f['title']} ({f['release_date']}) | Director: {f['director']} | RT score: {f['rt_score']}")
```

## 4. Convierte los datos a un DataFrame de pandas:

Muestra las primeras filas de DataFrame.

```python
df_films = pd.DataFrame(films)

for col in ["release_date", "running_time", "rt_score"]:
    df_films[col] = pd.to_numeric(df_films[col])

df_films[["title", "director", "release_date", "running_time", "rt_score"]].head()
```

# Nivel 3
API de Open Data Barcelona

Recursos:

- [Open Data BCN](https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS)

## 1. Utiliza la API de Open Data BCN para:

Buscar un dataset de tu interés mediante package_search y package_show. .

```python
CKAN_URL = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"

QUERY = "bicing"

resp = requests.get(f"{CKAN_URL}/package_search", params={"q": QUERY, "rows": 10}, timeout=20)
print("Código de estado:", resp.status_code)

resultado = resp.json()["result"]
print(f"Datasets encontrados: {resultado['count']}\n")

for pkg in resultado["results"]:
    formatos = sorted({r["format"] for r in pkg["resources"]})
    print(f"- {pkg['name']:<45} formatos: {formatos}")
```

## 2. De los resultados obtenidos, selecciona uno que tenga recursos disponibles en CSV o JSON.




```python
dataset_elegido, recurso_elegido = None, None

for pkg in resultado["results"]:
    for r in pkg["resources"]:
        if r["format"].upper() in ("CSV", "JSON") and r.get("datastore_active"):
            dataset_elegido, recurso_elegido = pkg["name"], r
            break
    if recurso_elegido:
        break

if recurso_elegido is None:
    raise ValueError("Ningún resultado tiene un recurso CSV/JSON con datastore activo. Prueba otra búsqueda (QUERY).")


resp = requests.get(f"{CKAN_URL}/package_show", params={"id": dataset_elegido}, timeout=20)
detalle = resp.json()["result"]

print("Dataset:", detalle["title"])
print("Recursos disponibles:")
for r in detalle["resources"]:
    print(f"  - {r['name'][:50]:<50} {r['format']:<5} datastore_active={r.get('datastore_active')}  id={r['id']}")

resource_id = recurso_elegido["id"]
print("\nRecurso seleccionado:", recurso_elegido["name"], "->", resource_id)
```

## 3. Con el resource_id del recurso seleccionado, realiza una consulta mediante datastore_search por:

Recuperar al menos 100 registros.

```python
resp = requests.get(
    f"{CKAN_URL}/datastore_search",
    params={"resource_id": resource_id, "limit": 100},
    timeout=30,
)
print("Código de estado:", resp.status_code)

res = resp.json()["result"]
registros = res["records"]
print("Registros totales en el recurso:", res.get("total"))
print("Registros recuperados:", len(registros))
```

## 4. Convierte los resultados a un DataFrame.

```python
df_bcn = pd.DataFrame(registros)
print(df_bcn.shape)
df_bcn.head()
```

## 5. Guarda DataFrame en un archivo .csv.

```python
df_bcn.to_csv("opendata_bcn.csv", index=False, encoding="utf-8")
print("Archivo guardado: opendata_bcn.csv")

pd.read_csv("opendata_bcn.csv").head()
```
